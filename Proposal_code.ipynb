```python
model = YOLO('yolov8l.pt')


results = model.train(
    data=os.path.join(DATASET_PATH, 'data.yaml'),
    epochs=100,
    patience=50,
    imgsz=640,
    batch=16,
    cos_lr=True,

    # --- Advanced Data Augmentation ---
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=15.0,
    translate=0.1,
    scale=0.5,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.1,
    copy_paste=0.1,

    # --- Project Naming ---
    project='runs/detect',
    name='traffic_signs_yolov8l_high_accuracy'
)

print("Training complete.")
```
